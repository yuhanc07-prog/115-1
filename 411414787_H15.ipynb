```python
data1=[]
data2=[]
data3=[]
while True:
  a=input("請輸入A倉文具，輸入'/'結束：")
  if a=="/":
    break
  data1.append(a)
while True:
  b=input("請輸入B倉文具，輸入'/'結束：")
  if b=="/":
    break
  data2.append(b)
data3=data1.copy()+data2.copy()
data3="/".join(data3)
print(f"二倉合併清單：\n{data3}")

```
