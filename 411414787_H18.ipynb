```python
commodity={"蘋果":50,"香蕉":30,"牛奶":60,"麵包":40}
a=[]
print("=== 超市商品 ===")
for i,j in commodity.items():
  print(f"{i}:{j}元")
while True:
  n=input("請輸入商品名稱（輸入 / 結束）：")
  if n=="/":
    break
  else:
    price=commodity[n]
    a.append(price)
print(f"總金額：{sum(a)}元")
```
