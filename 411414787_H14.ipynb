```python
def h(n):
  if n==0:
    return
  print(n)
  h(n-1)

h(5)
```

```python
def h(n):
  if n==1:
    return 1
  return n*h(n-1)

h(5)
```

```python
def f(n):
  if n==1:
    return 1
  if n==2:
    return 2
  return f(n-1)+f(n-2)

sum=f(10)
print(f"總共有{sum}種方式")
```
