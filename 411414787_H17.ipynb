```python
arms=["木棒","手槍","霰彈槍","狙擊槍","雷射劍"]
dam=(10,30,50,80,100)
for i,j in enumerate(arms):
  print(i,j)
while True:
  n=int(input("請輸入武器編號（0 結束）："))
  if n==0:
    break
  if 1<= n <= len(arms):
    print(arms[n-1],dam[n-1])
  else:
    print("找不到武器")
print("平均攻擊力：",sum(dam)/len(arms))
```
