```python
data=[]
while True:
  a=input("請輸入您要點的餐點，輸入'/'結束：")
  if a=="/":
    break
  data.append(a)
  if a=="沙拉":
    data.remove("沙拉")
    print("今日無沙拉，已自動移除。")
data.append("冰淇淋")
data=",".join(data)
print(f"完整訂單清單：\n{data}")
```
